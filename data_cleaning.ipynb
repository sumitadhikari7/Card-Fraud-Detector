```python
from sklearn.preprocessing import RobustScaler
```

```python
import pandas as pd
df = pd.read_csv("creditcard.csv")
```

```python

df['Hour'] = (df['Time']//3600)%24
```

```python
df['Hour'].unique()
```

```python
def get_time_risk(hour):
    if 2<= hour <=5:
        return 3 #High Risk
    if (hour > 5 and hour <= 8) or (hour >= 22):
        return 2 #Moderate Risk
    else:
        return 1 #Low Risk
df['Time Risk'] = df['Hour'].apply(get_time_risk)
```

```python
df['Amount'].describe()
```

```python
duplicates = df.duplicated().sum()
print(duplicates)
```

```python
df.drop_duplicates(inplace=True)
print(df.shape)
```

```python
# See how many $0 transactions are Fraud vs Normal
zero_amount_fraud = df[df['Amount'] == 0]['Class'].value_counts()
print("Breakdown of $0 transactions:")
print(zero_amount_fraud)
```

```python
rs = RobustScaler()
df['Scaled Amount'] = rs.fit_transform(df['Amount'].values.reshape(-1,1))
df['Is_Zero_Amount'] = (df['Amount'] == 0).astype(int)
```

```python
#drop amount, time, class and hour for training data
```

```python
df.to_csv("preprocessed_data.csv")
```
